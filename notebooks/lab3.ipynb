```python

import numpy as np


def standardise_loop(data):
    data = np.asarray(data, dtype=float)
    rows, cols = data.shape

    means = []
    stds = []

    for j in range(cols):
        column = [data[i, j] for i in range(rows)]
        mean = sum(column) / rows
        variance = sum((x - mean) ** 2 for x in column) / rows

        means.append(mean)
        stds.append(variance ** 0.5)

    result = []

    for i in range(rows):
        row = []
        for j in range(cols):
            if stds[j] == 0:
                row.append(0.0)
            else:
                row.append((data[i, j] - means[j]) / stds[j])
        result.append(row)

    return np.array(result)


def standardise_np(data):
    data = np.asarray(data, dtype=float)
    means = data.mean(axis=0)
    stds = data.std(axis=0)
    stds[stds == 0] = 1.0

    return (data - means) / stds
```

```python

rows = []

for N in [1_000, 100_000, 1_000_000]:
    py_list = list(range(N))
    np_arr = np.arange(N)

    n = 100 if N == 1_000 else 20

    t_loop = timeit.timeit(
        "loop_double(py_list)",
        globals=globals(),
        number=n
    ) / n

    t_np = timeit.timeit(
        "numpy_double(np_arr)",
        globals=globals(),
        number=n
    ) / n

    ratio = t_loop / t_np
    rows.append((N, t_loop, t_np, ratio))

print(f"{'N':>10} {'Loop(s)':>12} {'NumPy(s)':>12} {'Ratio':>10}")

for N, t_loop, t_np, ratio in rows:
    print(f"{N:>10} {t_loop:>12.6f} {t_np:>12.6f} {ratio:>10.2f}")
```

```python

import numpy as np
import time
import timeit
import tracemalloc

N = 1_000_000
py_list = list(range(N))
np_arr = np.arange(N)

def loop_double(values):
    result = []
    for x in values:
        result.append(x * 2)
    return result

def numpy_double(values):
    return values * 2

# perf_counter
for label, func, data in (
    ("loop", loop_double, py_list),
    ("numpy", numpy_double, np_arr)
):
    start = time.perf_counter()
    func(data)
    print(label, round(time.perf_counter() - start, 5), "seconds")

# timeit
n = 20
t_loop = timeit.timeit(
    "loop_double(py_list)", globals=globals(), number=n
) / n

t_np = timeit.timeit(
    "numpy_double(np_arr)", globals=globals(), number=n
) / n

print("loop", round(t_loop, 5),
      "numpy", round(t_np, 5),
      "ratio", round(t_loop / t_np, 2))

# Memory comparison
for label, make in (
    ("list", lambda: [x * 2 for x in range(N)]),
    ("array", lambda: np.arange(N) * 2)
):
    tracemalloc.start()
    obj = make()
    current, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()

    print(label,
          "current MB", round(current / 1e6, 1),
          "peak MB", round(peak / 1e6, 1))
```
